```python
df=df.drop_duplicates()
```

```python
df["director"]=df["director"].fillna("Unknown")

df["cast"]=df["cast"].fillna("Unknown")

df["country"]=df["country"].fillna("Unknown")

df["rating"]=df["rating"].fillna("Unknown")

df["date_added"]=df["date_added"].fillna("Unknown")

df["duration"]=df["duration"].fillna("Unknown")
```

```python
df["date_added"]=pd.to_datetime(df["date_added"],errors="coerce")
```

```python
df["year_added"]=df["date_added"].dt.year

df["month_added"]=df["date_added"].dt.month_name()
```

## Analytical Question 1
How has Netflix's catalogue evolved over time?

```python
release=df["release_year"].value_counts().sort_index()

fig=px.area(
    x=release.index,
    y=release.values,
    title="Growth of Netflix Content Over Time",
    labels={"x":"Release Year",
            "y":"Number of Titles"},
    color_discrete_sequence=["#E50914"]
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    font=dict(size=16)
)

fig.add_annotation(
    x=2018,
    y=max(release.values),
    text="Rapid growth after 2015",
    showarrow=True
)

fig.show()
```


Insight

Netflix experienced significant growth after 2015, reflecting rapid expansion of its global content catalogue.

## Analytical Question 2
What percentage of Netflix's catalogue consists of Movies and TV Shows?

```python
fig=px.pie(
    df,
    names="type",
    hole=.55,
    color="type",
    color_discrete_sequence=["#E50914","#221F1F"],
    title="Distribution of Netflix Content"
)

fig.update_layout(
    template="plotly_white",
    title_x=.5
)

fig.show()
```

Insight

Movies represent the majority of Netflix's catalogue, while TV Shows account for a smaller but substantial share.

## Analytical Question 3
How has the balance between Movies and TV Shows changed over time?

```python
trend=df.groupby(["release_year","type"]).size().reset_index(name="Titles")

fig=px.line(
    trend,
    x="release_year",
    y="Titles",
    color="type",
    markers=True,
    title="Movies vs TV Shows Over Time",
    color_discrete_sequence=["#E50914","#221F1F"]
)

fig.update_layout(
    template="plotly_white",
    title_x=.5
)

fig.show()
```

Insight

Movies consistently outnumber TV Shows, although both categories have grown over time.

## Analytical Question 4
Which countries contribute the most Netflix content?

```python
country=df.copy()

country["country"]=country["country"].str.split(",")

country=country.explode("country")

country["country"]=country["country"].str.strip()

top=country["country"].value_counts().head(15)

fig=px.bar(
    x=top.values,
    y=top.index,
    orientation="h",
    color=top.values,
    color_continuous_scale="Reds",
    title="Top 15 Countries Producing Netflix Content"
)

fig.update_layout(
    template="plotly_white",
    title_x=.5
)

fig.show()
```

Insight

The United States contributes the largest number of Netflix titles, followed by India and the United Kingdom.

## Analytical Question 5
How globally distributed is Netflix's catalogue?

```python
country_count=country["country"].value_counts().reset_index()

country_count.columns=["country","Titles"]

fig=px.choropleth(
    country_count,
    locations="country",
    locationmode="country names",
    color="Titles",
    hover_name="country",
    color_continuous_scale="Reds",
    title="Global Distribution of Netflix Content"
)

fig.update_layout(
    template="plotly_white",
    title_x=.5
)

fig.show()
```

Insight

Netflix content is concentrated in a few countries, with the United States and India contributing the largest share, while many other countries have much smaller catalogues.

## Analytical Question 6
Do countries specialize in Movies or TV Shows?

```python
country_type = df.copy()

country_type["country"] = country_type["country"].str.split(",")

country_type = country_type.explode("country")

country_type["country"] = country_type["country"].str.strip()

top15 = country_type["country"].value_counts().head(15).index

country_type = country_type[country_type["country"].isin(top15)]

group = country_type.groupby(["country","type"]).size().reset_index(name="Titles")

fig = px.bar(
    group,
    x="country",
    y="Titles",
    color="type",
    barmode="stack",
    title="Movie vs TV Show Production by Country",
    color_discrete_sequence=["#E50914","#221F1F"]
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Country",
    yaxis_title="Number of Titles"
)

fig.show()
```

Insight

Most countries produce significantly more Movies than TV Shows, although some have a relatively balanced catalogue.

## Analytical Question 7
Which genres dominate Netflix?

```python
genre = df.copy()

genre["listed_in"] = genre["listed_in"].str.split(",")

genre = genre.explode("listed_in")

genre["listed_in"] = genre["listed_in"].str.strip()

genre_count = genre.groupby(["type","listed_in"]).size().reset_index(name="Titles")

fig = px.sunburst(
    genre_count,
    path=["type","listed_in"],
    values="Titles",
    color="Titles",
    color_continuous_scale="Reds",
    title="Genre Distribution of Netflix Content"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

Insight

Drama, International Movies, and Comedy dominate Netflix's catalogue, while TV Shows occupy a smaller share across genres.

## Analytical Question 8
How do genres differ between Movies and TV Shows?

```python
heat = genre.groupby(["listed_in","type"]).size().reset_index(name="Count")

heat = heat.pivot(
    index="listed_in",
    columns="type",
    values="Count"
).fillna(0)

fig = px.imshow(
    heat,
    color_continuous_scale="Reds",
    title="Genre Distribution Across Content Types",
    aspect="auto"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

Insight

Some genres are strongly associated with Movies, while others are more common among TV Shows

## Analytical Question 9
Which countries have the greatest genre diversity?

```python
diversity = genre.copy()

diversity["country"] = diversity["country"].str.split(",")

diversity = diversity.explode("country")

diversity["country"] = diversity["country"].str.strip()

diversity = diversity.groupby(["country","listed_in"]).size().reset_index(name="Count")

country_genres = diversity.groupby("country")["listed_in"].nunique().reset_index()

country_genres.columns = ["country","Genres"]

top = country_genres.sort_values(
    by="Genres",
    ascending=False
).head(20)

fig = px.treemap(
    top,
    path=["country"],
    values="Genres",
    color="Genres",
    color_continuous_scale="Reds",
    title="Countries with the Greatest Genre Diversity"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

Insight

Countries with larger catalogues generally exhibit greater genre diversity, indicating broader content offerings.

## Analytical Question 10
How are audience ratings distributed across Movies and TV Shows?

```python
rating = df.groupby(["rating","type"]).size().reset_index(name="Count")

pivot = rating.pivot(
    index="rating",
    columns="type",
    values="Count"
).fillna(0)

fig = px.imshow(
    pivot,
    color_continuous_scale="Reds",
    title="Audience Ratings by Content Type",
    aspect="auto"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

Insight

TV-MA and TV-14 are the dominant audience ratings, particularly for Movies, reflecting Netflix's focus on mature and teen audiences.
